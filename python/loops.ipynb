loops are used to iterate over a sequence like (list tuple or string) or other iterble objects.
they allow you to execute a block of code multiple times with different values from the sequences


```python
#  for loop
fruits=["Apple","Mango","Banana"]
for i in fruits:  # i is temporary  variables  and fruits is lsit name
    print(i)
```

```python
n=int(input("Enter start number"))
m=int(input("Enter the end number"))
for i in range(n,m+1):
    print(i)
```

```python
course="Python Programming"
for i in course:
    print(i)
```

```python
# nested loops
size=["Small","medium","large"]
cloth=["Shirt","Pant"]
for i in size:
    for x in cloth:
        print(i,x)
```

 # while loop

while loops are used to repeatedly execute a block of code as long as specified condition remains true.
they are useful when the number of iteration is not known beforehand and depends on dynamic condition.

```python
num=10
while num>0:
    print(num)
    num-=1
```

```python
i = 1
sum = 0

n = int(input("Enter a number"))

while i <= n:
    print(i)
    sum += i
    i += 1

print("Sum =", sum)
```
