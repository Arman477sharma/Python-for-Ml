conditional statements are used to execute certian block of code based on specific condition.

```python
# if statement
temp = 25

if temp > 20:
    print("It's a warm day") # if first condition is true then first part is executed otherwise else part is executed
else:
    print("It's a cold day")
```

```python
# if elif and else
temp=18
if temp>30:
    print("It's a hot day")
elif temp>20:
    print("It's a warm day")
else:
    print("It's a cool day")
```

```python
# nested if if inside if
temp=30
weather ="sunny"
if temp > 15:
    if weather =="sunny":
        print("It's a sunny day and it's  warm")
else:
    print("It's a cool day")
```
