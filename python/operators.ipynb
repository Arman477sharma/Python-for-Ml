Arithmetic operators: handel mathematical calculation 
compariosn operators: evaluate and compare values
logical operators: combine boolean expersion 
assignment operators: manage the assignment and modification of variables values.


```python
# Arithmetic operators:
a=10
b=5
print(a+b)
print(a-b)
print(a*b)
print(a/b) # it give float value in new python version but when you add two // it give int value (floor value)
print(a//b)
print(a**b) # exponent
```

```python
# logical operators
a = True
b = False

print(a and b) # both should be true
print( a or b) # only one is true
print(not a) # opposite
```

```python
#  comparison operator
a=5
b=10
print(a==b)
print(a>=b)
print(a>b)
print(a<=b)
print(a<=b)
print(a!=b)
```

```python
# Assignment operator
a=5
a=a+1
print(a)

#  another way
a=5
a+=1
print(a)

```
