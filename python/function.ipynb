```python
# multiply two numbers
def multiply(a,b): # a, b are parameters (placeholders)
    return a*b

multiply(2,3) # 2,3 are arguments
```

Lambda map and zip function

```python
# Anonymous function
# lambda parameter : expression but only one
square=lambda x: x**2
square(5)
```

```python
# zip function
first_names = ["John", "Mary", "Bob"]
last_names = ["Smith", "Jane", "Smith"]
full_names = list(zip(first_names, last_names))
full_names
```

```python
# map
def square(x):
    return x**2
num=[2,3,4,5]
list(map(square,num))
```
