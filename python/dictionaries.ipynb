dictionaries are built in data strucutre un pyhton used to store key value pairs.
they are  ordered collection where each key is unique and is used to map to a value

```python
# creating a dictionaries

person={
    "name":"Arman Sharma", # name is key and its value is arman Sharma
    "Age":19,
    "RollNo":57144,
    "Height":5.6
}

print(person)
```

```python
type(person)
```

```python
# acess the elements
person["name"]
```

```python
person.keys() # all keys
```

```python
person["name"]="John" # replace an elements
print(person)
```

```python
person.values()  # print all values 
```
