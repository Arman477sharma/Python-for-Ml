List are used to collection of items which can be of any data type. list are mutable and can contain duplicates elements.

```python
# creating list
empty_list = []

# fruits_lsit
fruits = ["Apple","Mango", "Banana"]
print(fruits)

# Friends list
friends = ["Marry","john","Bob"]
print(friends)

numbers = [1,2,3,4,5]
print(numbers)  # smae for others 

# mix list
mix=["Apple",1,2.5,False]
print(mix)
```

```python
# access the items of list
numbers[0] # access the first elements because indexing start from 0 to n-1
fruits[2]
```

```python
# slicing 
friends[0:2] # 0 is start index and 2 is end index but 2 is excluded 
friends[0:3]
friends[:] # print whole list 
fruits[:2]
fruits[1:]
```

```python
# Methods
# add an elemnt in list
# friends.append("Mosh") # add the data in end of lsit
# print(friends)

friends.insert(1,"Karan")
print(friends)
```

```python
friends.remove("Mosh")
print(friends)
```

```python
friends.pop() # remove the last element
print(friends)
```

```python
# replace 
numbers[1]=10 # replace the element at indexes one with 10
print(numbers)
```
